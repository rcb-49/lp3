```python
def cost_function(x):
    return (x+3)**2
def gradient(x):
    return 2 * (x+3)
```

```python
# xnew = xold - learning_rate * (df/dx)
learning_rate = 0.1
x = 2
tolerance = 1e-6
max_iterations = 100
```

```python
for i in range(max_iterations):
    xnew = x - learning_rate * gradient(x)
    if( abs(xnew - x) < tolerance):
        x = xnew
        break
    x = xnew
```

```python
print("Minimum x :",x)
```

```python
print("Minimum y :",round(cost_function(x),8))
```

```python
print("Final gradient :",round(gradient(x),8))
```
